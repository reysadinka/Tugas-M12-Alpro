```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji

karyawan = [
    Karyawan("Jaja", 5000000),
    Karyawan("Jojo", 5500000),
    Karyawan("Joshua", 4500000),
    Karyawan("Jeje", 6500000),
    Karyawan("Juju", 5159900),
]

for k in karyawan:
    print(f"Karyawan {k.nama} memiliki gaji {k.gaji}")
```

```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji
        self.energi = 100

    def lembur(self):
        print(f"{self.nama} sedang lembur!")
        self.energi -= 30
        self.gaji += 200000

karyawan = [
    Karyawan("Jaja", 5000000),
    Karyawan("Jojo", 5500000),
    Karyawan("Joshua", 4500000),
    Karyawan("Jeje", 6500000),
    Karyawan("Juju", 5159900),
]

for k in karyawan:
    k.lembur()

for k in karyawan:
    print(f"Karyawan {k.nama} memiliki gaji {k.gaji} | Sisa energi: {k.energi}")
```

```python
class Barang:
    def __init__(self, nama, harga):
        self.nama = nama
        self.harga = harga

baju = Barang("Kaos Polos", 100000)
print(f"Harga awal {baju.nama}: {baju.harga}")

baju.harga = 75000
print(f"Harga terbaru {baju.nama} setelah diskon: {baju.harga}")
```

```python
class RekeningBank:
    def __init__(self, nama_pemilik, saldo=0):
        self.nama_pemilik = nama_pemilik
        self.saldo = saldo

    def menabung(self, nominal):
        self.saldo += nominal
        print(f"Berhasil menabung Rp{nominal}. Saldo sekarang: Rp{self.saldo}")

    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
        else:
            print("Maaf, saldo tidak mencukupi.")
            
rekening_saya = RekeningBank("Budi", 150000)
rekening_saya.tarik_tunai(50000)
rekening_saya.tarik_tunai(200000)
rekening_saya.menabung(300000)
```
